```python
# 1. Fungsi GOR
def calculate_gor(GasProd, OilProd):
    gor_value = GasProd / OilProd
    return gor_value

# 2. Panggil fungsi def
hasil_gor = calculate_gor(1500000, 1000)

# 3. Cetak hasilnya
print("Gas-Oil Ratio:", hasil_gor, "scf/bbl")
```

```python
# 1. fungsi menghitung RF
def calculate_rf(np, ioip):
    return np / ioip

# 2. Memasukkan data data lapangan dan memanggil fungsi
persentase_rf = calculate_rf(25000, 100000) * 100

# 3. Menampilkan hasil dalam bentuk persen (%)
print("Recovery Factor Lapangan:", persentase_rf, "%")
```

```python
import pandas as pd
df = pd.read_csv('oil and gas production.csv')
```

```python
# 1. Mengisi nilai-nilai yang hilang dengan 0
df.fillna(0, inplace=True)
```

```python
# 2. Mengonversi kolom 'Reporting Year' menjadi tipe data datetime
df['Reporting_Date'] = pd.to_datetime(df['Reporting Year'], format='%Y')
```

```python
# 3. Mengganti nama kolom agar lebih mudah digunakan
df = df.rename(columns={
    'Gas Produced, Mcf': 'Gas_Produced_Mcf',
    'Water Produced, bbl': 'Water_Produced_bbl',
    'Oil Produced, bbl': 'Oil_Produced_bbl',
    'New Georeferenced Column': 'Georeference'
})
```

```python
# 4. Membuat fitur baru 'Water_Cut'
# Menangani pembagian dengan nol jika Oil_Produced_bbl + Water_Produced_bbl adalah nol
df['Water_Cut'] = df.apply(lambda row: row['Water_Produced_bbl'] / (row['Oil_Produced_bbl'] + row['Water_Produced_bbl']) if (row['Oil_Produced_bbl'] + row['Water_Produced_bbl']) != 0 else np.nan, axis=1)
```

```python
import plotly.express as px
```

```python
annual_production = df.groupby('Reporting_Date')[['Oil_Produced_bbl', 'Gas_Produced_Mcf']].sum().reset_index()

fig_time_series = px.line(annual_production,
                          x='Reporting_Date',
                          y=['Oil_Produced_bbl', 'Gas_Produced_Mcf'],
                          title='Tren Produksi Minyak dan Gas per Tahun',
                          labels={'Reporting_Date': 'Tahun Pelaporan',
                                  'value': 'Jumlah Produksi',
                                  'variable': 'Jenis Produksi'})

fig_time_series.update_yaxes(title_text='Produksi Minyak (bbl) / Gas (Mcf)')
fig_time_series.update_layout(hovermode='x unified')
fig_time_series.show()
```

```python
top_wells = df[df['Oil_Produced_bbl'] > 0].groupby('Well Name')['Oil_Produced_bbl'].sum().nlargest(10).reset_index()

fig_bar = px.bar(top_wells,
                 x='Well Name',
                 y='Oil_Produced_bbl',
                 title='Top 10 Sumur berdasarkan Total Produksi Minyak',
                 labels={'Well Name': 'Nama Sumur', 'Oil_Produced_bbl': 'Total Produksi Minyak (bbl)'})
fig_bar.show()
```

```python
production_data = df[(df['Oil_Produced_bbl'] > 0) & (df['Gas_Produced_Mcf'] > 0)]

fig_scatter = px.scatter(production_data,
                         x='Oil_Produced_bbl',
                         y='Gas_Produced_Mcf',
                         color='Well Type Code',
                         hover_name='Well Name',
                         title='Hubungan antara Produksi Gas dan Minyak berdasarkan Jenis Sumur',
                         labels={'Oil_Produced_bbl': 'Produksi Minyak (bbl)', 'Gas_Produced_Mcf': 'Produksi Gas (Mcf)'})
fig_scatter.show()
```

```python
df_combined_production = df.copy()
df_combined_production['Gas_Produced_BOE'] = df_combined_production['Gas_Produced_Mcf'] * 0.178


df_combined_production_filtered = df_combined_production[
    (df_combined_production['Oil_Produced_bbl'] > 0) | (df_combined_production['Gas_Produced_BOE'] > 0)
].copy()


df_melted_combined = pd.melt(df_combined_production_filtered,
                               id_vars=['Reporting Year'],
                               value_vars=['Oil_Produced_bbl', 'Gas_Produced_BOE'],
                               var_name='Jenis Produksi',
                               value_name='Jumlah Produksi')


df_melted_combined['Jenis Produksi'] = df_melted_combined['Jenis Produksi'].replace({
    'Oil_Produced_bbl': 'Produksi Minyak (bbl)',
    'Gas_Produced_BOE': 'Produksi Gas (BOE)'})


fig_combined_boxplot = px.box(df_melted_combined,
                              x='Reporting Year',
                              y='Jumlah Produksi',
                              color='Jenis Produksi',
                              title='Distribusi Produksi Minyak (bbl) dan Gas (BOE) per Tahun',
                              labels={'Reporting Year': 'Tahun Pelaporan',
                                      'Jumlah Produksi': 'Jumlah Produksi (bbl/BOE)'})

fig_combined_boxplot.update_layout(xaxis_title='Tahun Pelaporan', yaxis_title='Jumlah Produksi (bbl/BOE)')
fig_combined_boxplot.show()
```
